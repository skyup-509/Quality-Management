# SMT 공정 개선
* SMT(Surface Mount Technology)란?

    PCB에 구멍을 내지않고 칩을 직접 부착
    
* 사용 데이터: [AI Hub SMT 공정 개선 멀티모달 데이터](https://aihub.or.kr/aihubdata/data/view.do?pageIndex=1&currMenu=115&topMenu=100&srchOneDataTy=DATA006&srchOptnCnd=OPTNCND001&searchKeyword=&srchDetailCnd=DETAILCND001&srchOrder=ORDER001&srchPagePer=20&srchDataRealmCode=REALM012&aihubDataSe=data&dataSetSn=71900)

* SMT 공정 프로세스 설명: [(주)에이티에스로의 SMT 공정 절차 설명](https://blog.naver.com/atsro/221750964225)
    1. PCB 자동 로딩
    2. Solder Printing: 솔더를 PCB 표면상에 고르게 도포
       
       영향: 습도, 진동

    3. SPI(납'도포도'검사): 솔더가 고르게 잘 퍼졌는지 검사
    4. Chip Mounter: 부품을 흡착하여 PCB 위에 장착

       영향: 속도 $\rightarrow$ 저/중속 마운터 or 고속 마운터

    5. Reflow Soldering: 마운팅이 완료된 솔더를 경화시킴

       온도를 올리며 경화시킴.

       영행: 온도 (시간-구간마다 온도를 달리함)

    6. Unloading: PCB 적재 및 수납
    7. AOI(자동광학검사): 결함(외관) 검사 $\rightarrow$ 불량 시, 후공정

```python
def get_file_pairs(sensor_dir, image_dir):
    sensor_files = sorted([f for f in os.listdir(sensor_dir) if f.endswith('.xlsx')])
    pairs = []
    for sensor_file in sensor_files:
        base_name = sensor_file.replace('.xlsx', '')
        image_file = base_name + '.JPG'
        image_path = os.path.join(image_dir, image_file)
        sensor_path = os.path.join(sensor_dir, sensor_file)
        
        if os.path.exists(image_path):
            pairs.append({
                'base_name': base_name,
                'sensor_path': sensor_path,
                'image_path': image_path
            })
    return pairs

defect_pairs = get_file_pairs(SENSOR_DATA_DIR['defect'], IMAGE_DATA_DIR['defect'])
normal_pairs = get_file_pairs(SENSOR_DATA_DIR['normal'], IMAGE_DATA_DIR['normal'])
```

```python
import json
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import numpy as np
import os
import pandas as pd
from collections import Counter
from pathlib import Path

# 데이터 로드
DATA_ROOT = "./smt_data"
LABEL_DIR = Path(DATA_ROOT) / "Sample/02.라벨링데이터/사전공정/불량/1.미납"
IMAGE_DIR = Path(DATA_ROOT) / "Sample/01.원천데이터/이미지데이터/사전공정/불량/1.미납"

json_files = sorted(LABEL_DIR.glob("*.json"))
SAMPLE_JSON = json_files[1] # 사진 파일 선택

base_name = SAMPLE_JSON.stem
SAMPLE_IMAGE = IMAGE_DIR / f"{base_name}.JPG"

print(f"파일: {SAMPLE_JSON.name}\n")

with open(SAMPLE_JSON, 'r', encoding='utf-8') as f:
    data = json.load(f)

categories = {cat['id']: cat['name'] for cat in data['categories']}
img = Image.open(SAMPLE_IMAGE)
img_array = np.array(img)
img_width, img_height = img.size

defect_types = [categories[ann['category_id']] for ann in data['annotations'] 
               if '불량' in categories[ann['category_id']]]
main_defect = Counter(defect_types).most_common(1)[0][0] if defect_types else "결함 없음" # 불량 모드가 여러개 있을건데, 이 중 다수를 이미지의 최종 불량 모드로 선정

print(f"주요 결함: {main_defect}")
print(f"결함 개수: {len(defect_types)}개\n")

fig, ax = plt.subplots(1, 1, figsize=(12, 10))
ax.imshow(img_array)

for annotation in data['annotations']:
    x, y, w, h = annotation['bbox']
    category_id = annotation['category_id']
    category_name = categories[category_id]
    
    rect = patches.Rectangle((x, y), w, h, linewidth=2.5, 
                             edgecolor='red', facecolor='none', alpha=0.8)
    ax.add_patch(rect)
    
    text = f"ID:{category_id} ({category_name})"
    ax.text(x, y-8, text, fontsize=10, color='white', fontweight='bold',
           bbox=dict(boxstyle='round,pad=0.5', facecolor='red', alpha=0.8))

ax.set_title(f"불량 샘플 분석 (주요 결함: {main_defect})", 
            fontsize=14, fontweight='bold', pad=20)
ax.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

sensor_df = pd.DataFrame(data['sensor_data'][0]['sensor_sequence'])
display(sensor_df.round(4))

print("\n센서 통계:")
for col in ['temperature', 'humidity', 'vibration', 'acceleration', 'noise']:
    print(f"  {col}: {sensor_df[col].mean():.4f} (범위: {sensor_df[col].min():.4f}~{sensor_df[col].max():.4f})")
```

```python
import json
import pandas as pd
from pathlib import Path
from collections import Counter

DATA_ROOT = "./smt_data"
LABEL_DIR = Path(DATA_ROOT) / "Sample/02.라벨링데이터/사전공정/불량/1.미납"

json_files = sorted(LABEL_DIR.glob("*.json"))
categories = None
all_defects = {}
all_sensor_data = {}

for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    if not categories:
        categories = {cat['id']: cat['name'] for cat in data['categories']}
    
    # 모든 annotation 중 불량_* 만 필터링
    defect_types = [categories[ann['category_id']] for ann in data['annotations'] 
                   if '불량' in categories[ann['category_id']]]
    
    # 불량이 없으면 스킵
    if not defect_types:
        continue
    
    main_defect = Counter(defect_types).most_common(1)[0][0]
    sensor_df = pd.DataFrame(data['sensor_data'][0]['sensor_sequence'])
    
    if main_defect not in all_defects:
        all_defects[main_defect] = {
            'temps': [], 'humids': [], 'vibs': [], 'noises': [], 'count': 0
        }
        all_sensor_data[main_defect] = {
            'temps': [], 'humids': [], 'vibs': [], 'noises': []
        }
    
    all_defects[main_defect]['temps'].append(sensor_df['temperature'].mean())
    all_defects[main_defect]['humids'].append(sensor_df['humidity'].mean())
    all_defects[main_defect]['vibs'].append(sensor_df['vibration'].mean())
    all_defects[main_defect]['noises'].append(sensor_df['noise'].mean())
    all_defects[main_defect]['count'] += 1

    all_sensor_data[main_defect]['temps'].extend(sensor_df['temperature'].tolist())
    all_sensor_data[main_defect]['humids'].extend(sensor_df['humidity'].tolist())
    all_sensor_data[main_defect]['vibs'].extend(sensor_df['vibration'].tolist())
    all_sensor_data[main_defect]['noises'].extend(sensor_df['noise'].tolist())

for defect_type in sorted(all_defects.keys()):
    stats = all_defects[defect_type]
    print(f"{defect_type} (n={stats['count']})")
    print(f"  온도: {min(stats['temps']):.2f}~{max(stats['temps']):.2f}°C")
    print(f"  습도: {min(stats['humids']):.2f}~{max(stats['humids']):.2f}%")
    print(f"  진동: {min(stats['vibs']):.4f}~{max(stats['vibs']):.4f}m/s²")
    print(f"  노이즈: {min(stats['noises']):.2f}~{max(stats['noises']):.2f}dB")
    print()

# 분포 시각화
defect_types = sorted(all_defects.keys())
sensor_names = ['temps', 'humids', 'vibs', 'noises']
sensor_labels = ['Temperature (°C)', 'Humidity (%)', 'Vibration (m/s²)', 'Noise (dB)']

fig, axes = plt.subplots(len(defect_types), 4, figsize=(16, 4 * len(defect_types)))
if len(defect_types) == 1:
    axes = axes.reshape(1, -1)
 
for idx, defect_type in enumerate(defect_types):
    for sensor_idx, (sensor_name, sensor_label) in enumerate(zip(sensor_names, sensor_labels)):
        ax = axes[idx, sensor_idx]
        data = all_sensor_data[defect_type][sensor_name]
        
        ax.hist(data, bins=30, alpha=0.7, color='steelred', edgecolor='black')
        ax.set_title(f'{defect_type}\n{sensor_label}')
        ax.set_xlabel(sensor_label)
        ax.set_ylabel('Frequency')
        ax.grid(axis='y', alpha=0.3)
        
        # 평균선 추가
        mean_val = np.mean(data)
        ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}')
        ax.legend()

plt.tight_layout()
```

```python
import json
import pandas as pd
from pathlib import Path

DATA_ROOT = "./smt_data"
NORMAL_LABEL_DIR = Path(DATA_ROOT) / "Sample/02.라벨링데이터/사전공정/정상/1.미납"

json_files = sorted(NORMAL_LABEL_DIR.glob("*.json"))

normal_sensors = {
    'temps': [], 'humids': [], 'vibs': [], 'accels': [], 'noises': []
}

for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    sensor_df = pd.DataFrame(data['sensor_data'][0]['sensor_sequence'])
    
    normal_sensors['temps'].append(sensor_df['temperature'].mean())
    normal_sensors['humids'].append(sensor_df['humidity'].mean())
    normal_sensors['vibs'].append(sensor_df['vibration'].mean())
    normal_sensors['accels'].append(sensor_df['acceleration'].mean())
    normal_sensors['noises'].append(sensor_df['noise'].mean())

print(f"온도: {min(normal_sensors['temps']):.2f}~{max(normal_sensors['temps']):.2f}°C (평균: {np.mean(normal_sensors['temps']):.2f})")
print(f"습도: {min(normal_sensors['humids']):.2f}~{max(normal_sensors['humids']):.2f}% (평균: {np.mean(normal_sensors['humids']):.2f})")
print(f"진동: {min(normal_sensors['vibs']):.4f}~{max(normal_sensors['vibs']):.4f}m/s² (평균: {np.mean(normal_sensors['vibs']):.4f})")
print(f"가속도: {min(normal_sensors['accels']):.4f}~{max(normal_sensors['accels']):.4f}g (평균: {np.mean(normal_sensors['accels']):.4f})")
print(f"노이즈: {min(normal_sensors['noises']):.2f}~{max(normal_sensors['noises']):.2f}dB (평균: {np.mean(normal_sensors['noises']):.2f})")

# 분포 시각화
sensor_names = ['temps', 'humids', 'vibs', 'accels', 'noises']
sensor_labels = ['Temperature (°C)', 'Humidity (%)', 'Vibration (m/s²)', 'Acceleration (g)', 'Noise (dB)']

fig, axes = plt.subplots(1, 5, figsize=(18, 4))
 
for sensor_idx, (sensor_name, sensor_label) in enumerate(zip(sensor_names, sensor_labels)):
    ax = axes[sensor_idx]
    data = normal_sensors[sensor_name]
    
    ax.hist(data, bins=30, alpha=0.7, color='green', edgecolor='black')
    ax.set_title(f'Normal Data\n{sensor_label}')
    ax.set_xlabel(sensor_label)
    ax.set_ylabel('Frequency')
    ax.grid(axis='y', alpha=0.3)
    
    # 평균선 추가
    mean_val = np.mean(data)
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}')
    ax.legend()
 
plt.tight_layout()
```

```python
import json
import pandas as pd
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import matplotlib as mpl

# 한글 폰트 설정
mpl.rcParams['font.family'] = 'Malgun Gothic'  # 맑은 고딕
mpl.rcParams['axes.unicode_minus'] = False     # 마이너스 깨짐 방지

DATA_ROOT = "./smt_data"

# ===== 1. 정상 데이터 처리 =====
print("=" * 80)
print("1단계: 정상 데이터 분석 중...")
print("=" * 80)

NORMAL_LABEL_DIR = Path(DATA_ROOT) / "Sample/02.라벨링데이터/사전공정/정상/1.미납"
json_files = sorted(NORMAL_LABEL_DIR.glob("*.json"))

normal_sensors = {
    'temps': [], 'humids': [], 'vibs': [], 'accels': [], 'noises': []
}

for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    sensor_df = pd.DataFrame(data['sensor_data'][0]['sensor_sequence'])
    
    normal_sensors['temps'].append(sensor_df['temperature'].mean())
    normal_sensors['humids'].append(sensor_df['humidity'].mean())
    normal_sensors['vibs'].append(sensor_df['vibration'].mean())
    normal_sensors['accels'].append(sensor_df['acceleration'].mean())
    normal_sensors['noises'].append(sensor_df['noise'].mean())

# 정상 데이터 범위 저장
normal_limits = {
    'temps': {'min': min(normal_sensors['temps']), 'max': max(normal_sensors['temps']), 'mean': np.mean(normal_sensors['temps'])},
    'humids': {'min': min(normal_sensors['humids']), 'max': max(normal_sensors['humids']), 'mean': np.mean(normal_sensors['humids'])},
    'vibs': {'min': min(normal_sensors['vibs']), 'max': max(normal_sensors['vibs']), 'mean': np.mean(normal_sensors['vibs'])},
    'accels': {'min': min(normal_sensors['accels']), 'max': max(normal_sensors['accels']), 'mean': np.mean(normal_sensors['accels'])},
    'noises': {'min': min(normal_sensors['noises']), 'max': max(normal_sensors['noises']), 'mean': np.mean(normal_sensors['noises'])}
}

print("✓ 정상 데이터 범위 (LSL ~ USL):")
print(f"  온도: {normal_limits['temps']['min']:.2f}~{normal_limits['temps']['max']:.2f}°C")
print(f"  습도: {normal_limits['humids']['min']:.2f}~{normal_limits['humids']['max']:.2f}%")
print(f"  진동: {normal_limits['vibs']['min']:.4f}~{normal_limits['vibs']['max']:.4f}m/s²")
print(f"  가속도: {normal_limits['accels']['min']:.4f}~{normal_limits['accels']['max']:.4f}g")
print(f"  노이즈: {normal_limits['noises']['min']:.2f}~{normal_limits['noises']['max']:.2f}dB")

# ===== 2. 불량 데이터 처리 =====
print("\n" + "=" * 80)
print("2단계: 불량 데이터 분석 중...")
print("=" * 80)

DEFECT_LABEL_DIR = Path(DATA_ROOT) / "Sample/02.라벨링데이터/사전공정/불량/1.미납"
json_files = sorted(DEFECT_LABEL_DIR.glob("*.json"))

categories = None
all_defects = {}
all_sensor_data = {}

for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    if not categories:
        categories = {cat['id']: cat['name'] for cat in data['categories']}
    
    defect_types = [categories[ann['category_id']] for ann in data['annotations'] 
                   if '불량' in categories[ann['category_id']]]
    
    if not defect_types:
        continue
    
    main_defect = Counter(defect_types).most_common(1)[0][0]
    sensor_df = pd.DataFrame(data['sensor_data'][0]['sensor_sequence'])
    
    if main_defect not in all_defects:
        all_defects[main_defect] = {
            'temps': [], 'humids': [], 'vibs': [], 'accels': [], 'noises': [], 'count': 0
        }
        all_sensor_data[main_defect] = {
            'temps': [], 'humids': [], 'vibs': [], 'accels': [], 'noises': []
        }
    
    all_defects[main_defect]['temps'].append(sensor_df['temperature'].mean())
    all_defects[main_defect]['humids'].append(sensor_df['humidity'].mean())
    all_defects[main_defect]['vibs'].append(sensor_df['vibration'].mean())
    all_defects[main_defect]['accels'].append(sensor_df['acceleration'].mean())
    all_defects[main_defect]['noises'].append(sensor_df['noise'].mean())
    all_defects[main_defect]['count'] += 1
    
    all_sensor_data[main_defect]['temps'].extend(sensor_df['temperature'].tolist())
    all_sensor_data[main_defect]['humids'].extend(sensor_df['humidity'].tolist())
    all_sensor_data[main_defect]['vibs'].extend(sensor_df['vibration'].tolist())
    all_sensor_data[main_defect]['accels'].extend(sensor_df['acceleration'].tolist())
    all_sensor_data[main_defect]['noises'].extend(sensor_df['noise'].tolist())

print(f"✓ {len(all_defects)}개의 불량 타입 발견")

# ===== 3. 통계 출력 =====
print("\n" + "=" * 80)
print("정상 데이터 통계")
print("=" * 80)
print(f"온도: {normal_limits['temps']['min']:.2f}~{normal_limits['temps']['max']:.2f}°C (평균: {normal_limits['temps']['mean']:.2f})")
print(f"습도: {normal_limits['humids']['min']:.2f}~{normal_limits['humids']['max']:.2f}% (평균: {normal_limits['humids']['mean']:.2f})")
print(f"진동: {normal_limits['vibs']['min']:.4f}~{normal_limits['vibs']['max']:.4f}m/s² (평균: {normal_limits['vibs']['mean']:.4f})")
print(f"가속도: {normal_limits['accels']['min']:.4f}~{normal_limits['accels']['max']:.4f}g (평균: {normal_limits['accels']['mean']:.4f})")
print(f"노이즈: {normal_limits['noises']['min']:.2f}~{normal_limits['noises']['max']:.2f}dB (평균: {normal_limits['noises']['mean']:.2f})")

print("\n" + "=" * 80)
print("불량 데이터 통계")
print("=" * 80)
for defect_type in sorted(all_defects.keys()):
    stats = all_defects[defect_type]
    print(f"\n{defect_type} (n={stats['count']})")
    print(f"  온도: {min(stats['temps']):.2f}~{max(stats['temps']):.2f}°C")
    print(f"  습도: {min(stats['humids']):.2f}~{max(stats['humids']):.2f}%")
    print(f"  진동: {min(stats['vibs']):.4f}~{max(stats['vibs']):.4f}m/s²")
    print(f"  가속도: {min(stats['accels']):.4f}~{max(stats['accels']):.4f}g")
    print(f"  노이즈: {min(stats['noises']):.2f}~{max(stats['noises']):.2f}dB")

# ===== 4. 시각화 설정 =====
defect_types = sorted(all_defects.keys())
sensor_names = ['temps', 'humids', 'vibs', 'accels', 'noises']
sensor_labels = ['Temperature (°C)', 'Humidity (%)', 'Vibration (m/s²)', 'Acceleration (g)', 'Noise (dB)']

# ===== 5. 불량별 히스토그램 (정상 범위 표시) =====

fig, axes = plt.subplots(len(defect_types), 5, figsize=(20, 4 * len(defect_types)))
if len(defect_types) == 1:
    axes = axes.reshape(1, -1)

for idx, defect_type in enumerate(defect_types):
    for sensor_idx, (sensor_name, sensor_label) in enumerate(zip(sensor_names, sensor_labels)):
        ax = axes[idx, sensor_idx]
        data = all_sensor_data[defect_type][sensor_name]
        
        # 히스토그램
        ax.hist(data, bins=30, alpha=0.7, color='steelblue', edgecolor='black')
        ax.set_title(f'{defect_type}\n{sensor_label}')
        ax.set_xlabel(sensor_label)
        ax.set_ylabel('Frequency')
        ax.grid(axis='y', alpha=0.3)
        
        # 정상 범위 (USL, LSL) 표시 - 빨간색 수직선
        lsl = normal_limits[sensor_name]['min']
        usl = normal_limits[sensor_name]['max']
        ax.axvline(lsl, color='red', linestyle='--', linewidth=2, label=f'LSL: {lsl:.2f}')
        ax.axvline(usl, color='red', linestyle='--', linewidth=2, label=f'USL: {usl:.2f}')
        
        # 불량 데이터 평균선 - 초록색
        mean_val = np.mean(data)
        ax.axvline(mean_val, color='green', linestyle='-', linewidth=2, label=f'Mean: {mean_val:.2f}')
        
        ax.legend(fontsize=8)

plt.tight_layout()
plt.show()

# ===== 6. 박스플롯 비교 (정상 vs 불량) =====

fig, axes = plt.subplots(1, 5, figsize=(20, 6))

for sensor_idx, (sensor_name, sensor_label) in enumerate(zip(sensor_names, sensor_labels)):
    ax = axes[sensor_idx]
    
    # 데이터 준비
    data_for_box = []
    labels_list = []
    
    # 정상 데이터
    data_for_box.append(normal_sensors[sensor_name])
    labels_list.append('Normal')
    
    # 불량 데이터
    for defect_type in defect_types:
        data_for_box.append(all_sensor_data[defect_type][sensor_name])
        labels_list.append(defect_type.replace('불량_', ''))
    
    bp = ax.boxplot(data_for_box, labels=labels_list, patch_artist=True)
    
    # 색상 설정
    bp['boxes'][0].set_facecolor('lightgreen')  # 정상은 초록색
    for i in range(1, len(bp['boxes'])):
        bp['boxes'][i].set_facecolor('lightcoral')  # 불량은 빨간색
    
    # 정상 범위 표시 (빨간색 음영)
    lsl = normal_limits[sensor_name]['min']
    usl = normal_limits[sensor_name]['max']
    ax.axhspan(lsl, usl, alpha=0.2, color='red', label='Normal Range')
    
    ax.set_title(sensor_label, fontsize=12, fontweight='bold')
    ax.set_ylabel(sensor_label)
    ax.grid(axis='y', alpha=0.3)
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=45, ha='right')
    ax.legend()

plt.tight_layout()
plt.show()

# ===== 7. 바이올린 플롯 비교 =====

fig, axes = plt.subplots(1, 5, figsize=(20, 6))

for sensor_idx, (sensor_name, sensor_label) in enumerate(zip(sensor_names, sensor_labels)):
    ax = axes[sensor_idx]
    
    # 데이터 준비
    plot_data = []
    labels_list = []
    colors_list = []
    
    # 정상 데이터
    plot_data.extend(normal_sensors[sensor_name])
    labels_list.extend(['Normal'] * len(normal_sensors[sensor_name]))
    colors_list.extend(['green'] * len(normal_sensors[sensor_name]))
    
    # 불량 데이터
    for defect_type in defect_types:
        plot_data.extend(all_sensor_data[defect_type][sensor_name])
        labels_list.extend([defect_type.replace('불량_', '')] * len(all_sensor_data[defect_type][sensor_name]))
        colors_list.extend(['red'] * len(all_sensor_data[defect_type][sensor_name]))
    
    df_plot = pd.DataFrame({
        'value': plot_data,
        'type': labels_list
    })
    
    sns.violinplot(data=df_plot, x='type', y='value', ax=ax, palette=['lightgreen'] + ['lightcoral'] * len(defect_types))
    
    # 정상 범위 표시 - 빨간색
    lsl = normal_limits[sensor_name]['min']
    usl = normal_limits[sensor_name]['max']
    ax.axhspan(lsl, usl, alpha=0.2, color='red')
    
    ax.set_title(sensor_label, fontsize=12, fontweight='bold')
    ax.set_xlabel('Type')
    ax.set_ylabel(sensor_label)
    ax.grid(axis='y', alpha=0.3)
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=45, ha='right')

plt.tight_layout()
plt.show()

# ===== 8. 상세 통계 =====
print("\n" + "=" * 80)
print("상세 통계")
print("=" * 80)

print("\n[정상 데이터]")
for sensor_name, sensor_label in zip(sensor_names, sensor_labels):
    data = normal_sensors[sensor_name]
    print(f"  {sensor_label}")
    print(f"    Mean: {np.mean(data):.4f}, Std: {np.std(data):.4f}")
    print(f"    Min: {np.min(data):.4f}, Max: {np.max(data):.4f}")
    print(f"    Median: {np.median(data):.4f}")

for defect_type in defect_types:
    print(f"\n[{defect_type}]")
    for sensor_name, sensor_label in zip(sensor_names, sensor_labels):
        data = all_sensor_data[defect_type][sensor_name]
        print(f"  {sensor_label}")
        print(f"    Mean: {np.mean(data):.4f}, Std: {np.std(data):.4f}")
        print(f"    Min: {np.min(data):.4f}, Max: {np.max(data):.4f}")

print("\n" + "=" * 80)
print("✓ 모든 분석 완료!")
print("=" * 80)
```
